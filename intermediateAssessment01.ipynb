orders.


```python
#Highest total quantity sold
item_quantity={}
for a in sales_data:
  item_quantity[a["item"]]=item_quantity.get(a["item"],0)+ a["quantity"]
  highest_sold_item= max(item_quantity,key=item_quantity.get)
print("Highest sold item: ",highest_sold_item ,"\nQuantity: ", item_quantity[highest_sold_item])
```

3.1 Write a function apply_discount(orders, discount_rate) that takes
a list of orders and applies a percentage discount (e.g., 10% = 0.10) to
the price_per_item for each order.
Return a new list of orders (do not modify the original list).

```python
#Discount

def apply_discount(orders, discount_rate):
    discounted_orders = []

    for order in orders:
        new_order = order.copy()   # create a shallow copy
        new_order["Price_Per_item"] = order["Price_Per_item"] * (1 - discount_rate)
        discounted_orders.append(new_order)

    return discounted_orders
discounted_sales = apply_discount(sales_data, 0.10)

print("\nOriginal Orders:")
print(sales_data)

print("\nDiscounted Orders:")
print(discounted_sales)
```

4.1 Write a custom iterator class OrderIterator to iterate through the
orders, yielding one order at a time.


```python

class OrderIterator:
    def __init__(self, orders):
        self.orders = orders
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.orders):
            raise StopIteration
        order = self.orders[self.index]
        self.index += 1
        return order
orders = [
    {"order_id": 1, "customer": "Sankar"},
    {"order_id": 2, "customer": "Saritha"},
    {"order_id": 3, "customer": "Saranya"}
]

order_iterator = OrderIterator(orders)

for order in order_iterator:
    print(order)
```

4.2 Create a dictionary grouping all orders by customer using a
dictionary comprehension.

```python
orders_by_customer= {
    customer: [a for a in sales_data if a["customer"]==customer]
    for customer in set(a["customer"] for a in sales_data)
    }
print(orders_by_customer)
```
