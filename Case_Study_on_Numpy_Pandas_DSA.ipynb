Convert the mpg column into a NumPy array and calculate:
○ The mean, median, and standard deviation of mpg.
○ The number of cars with mpg greater than 25.

Using NumPy, filter all cars with more than 6 cylinders.
Return the corresponding car_name as a list.

Compute the 25th, 50th, and 75th percentiles of the weight column using NumPy.
4. Array Manipulation
Convert the acceleration column into a NumPy array and normalize its values
(scale between 0 and 1).

Increase all horsepower values by 10% and store the updated values in a new
NumPy array. Handle missing data (if any) by replacing it with the mean of the
column before applying the increase.
6. Boolean Indexing
Find the average displacement of cars with an origin of 2 (Europe) using NumPy
indexing.
7. Matrix Operations
Create a 2D NumPy array containing the columns mpg, horsepower, and weight.
Compute the dot product of this matrix with a given vector [1, 0.5, -0.2].

f
ive car names.

Calculate the mean mpg for cars grouped by the number of cylinders using NumPy

```python
import pandas as pd
import numpy as np

# Load the dataset
# Handling '?' as NaN during loading helps with numerical operations
df = pd.read_csv('auto-mpg.csv', na_values='?')
mpg_array = df['mpg'].to_numpy()

# Mean, median, and standard deviation
mean_mpg = np.mean(mpg_array)
median_mpg = np.median(mpg_array)
std_mpg = np.std(mpg_array)

# Number of cars with mpg > 25
count_gt_25 = np.sum(mpg_array > 25)

print(f"Mean: {mean_mpg}, Median: {median_mpg}, Std Dev: {std_mpg}")
print(f"Cars with mpg > 25: {count_gt_25}")
# Using boolean indexing in NumPy
more_than_6_cyl = df[df['cylinders'].to_numpy() > 6]['car name'].tolist()
print(f"Cars with > 6 cylinders: {more_than_6_cyl[:5]}... (Total: {len(more_than_6_cyl)})")
weight_array = df['weight'].to_numpy()
percentiles = np.percentile(weight_array, [25, 50, 75])
print(f"25th, 50th, and 75th percentiles of weight: {percentiles}")
acc_array = df['acceleration'].to_numpy()
norm_acc = (acc_array - np.min(acc_array)) / (np.max(acc_array) - np.min(acc_array))
print(f"Normalized acceleration (first 5): {norm_acc[:5]}")
hp_array = df['horsepower'].to_numpy()

# Replace NaNs with the mean
mean_hp = np.nanmean(hp_array)
hp_filled = np.where(np.isnan(hp_array), mean_hp, hp_array)

# Increase by 10%
hp_updated = hp_filled * 1.10
print(f"Updated Horsepower (first 5): {hp_updated[:5]}")
origin_array = df['origin'].to_numpy()
disp_array = df['displacement'].to_numpy()

avg_disp_europe = np.mean(disp_array[origin_array == 2])
print(f"Average displacement for Europe (Origin 2): {avg_disp_europe}")
# Fill missing horsepower for matrix operations
df_filled = df.copy()
df_filled['horsepower'] = df_filled['horsepower'].fillna(df_filled['horsepower'].mean())

matrix_2d = df_filled[['mpg', 'horsepower', 'weight']].to_numpy()
vector = np.array([1, 0.5, -0.2])

dot_product = np.dot(matrix_2d, vector)
print(f"Dot product result (first 5): {dot_product[:5]}")
# Get indices that would sort the array
sorted_indices = np.argsort(df['model year'].to_numpy())[::-1]
top_5_names = df['car name'].iloc[sorted_indices[:5]].tolist()
print(f"Top 5 car names (newest first): {top_5_names}")
correlation = np.corrcoef(df['mpg'], df['weight'])[0, 1]
print(f"Correlation between mpg and weight: {correlation}")
cyls = df['cylinders'].unique()
mean_mpg_per_cyl = {cyl: np.mean(df[df['cylinders'] == cyl]['mpg']) for cyl in cyls}
print(f"Mean MPG by cylinder: {mean_mpg_per_cyl}")
```

Load the dataset into a Pandas DataFrame. Display:

2. Filtering and Indexing
Find all cars manufactured in 1975 with a weight less than 3000. Return the
DataFrame with selected columns: car_name, weight, and mpg.
3. Handling Missing Data
Identify if there are any missing values in the dataset. Replace missing values in the
horsepower column with the column's median.
4. Data Transformation
Add a new column power_to_weight_ratio, calculated as horsepower / weight.
5. Group By
Group the cars by origin and calculate the mean mpg for each group.

Sort the DataFrame by mpg in descending order and display the top 10 cars with

Apply Function
Create a new column performance_score using a custom function:
def performance_score(row):

7. Apply this function to each row and store the result in the new column.
8. Visualization Preparation
Generate a summary DataFrame with:
○ Average mpg, weight, and horsepower for each model_year.
9. Exporting Data
Save a subset of the data containing only mpg, cylinders, horsepower, and weight
for cars with mpg > 30 into a CSV file named high_mpg_cars.csv.
10. Finding Anomalies
Identify potential outliers in the mpg column using the Interquartile Range (IQR)
method. Specifically:
● Calculate the IQR for mpg.
2
● Define outliers as values less than Q1 - 1.5 * IQR or greater than Q3 +
1.5 * IQR.
● Create a DataFrame of cars classified as outliers, displaying car_name, mpg,
and model_year.

```python
print("First 10 rows:")
print(df.head(10))

print(f"\nTotal rows and columns: {df.shape}")

print("\nSummary Statistics:")
print(df.describe())
filtered_cars = df[(df['model year'] == 75) & (df['weight'] < 3000)]
print(filtered_cars[['car name', 'weight', 'mpg']])
print(f"Missing values count:\n{df.isnull().sum()}")

df['horsepower'] = df['horsepower'].fillna(df['horsepower'].median())
df['power_to_weight_ratio'] = df['horsepower'] / df['weight']
mean_mpg_origin = df.groupby('origin')['mpg'].mean()
print(mean_mpg_origin)
top_10_mpg = df.sort_values(by='mpg', ascending=False).head(10)
print(top_10_mpg)
def performance_score(row):
    return row['mpg'] * row['acceleration'] / row['weight']

df['performance_score'] = df.apply(performance_score, axis=1)
summary_df = df.groupby('model year')[['mpg', 'weight', 'horsepower']].mean()
print(summary_df)
high_mpg_cars = df[df['mpg'] > 30][['mpg', 'cylinders', 'horsepower', 'weight']]
high_mpg_cars.to_csv('high_mpg_cars.csv', index=False)
Q1 = df['mpg'].quantile(0.25)
Q3 = df['mpg'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['mpg'] < lower_bound) | (df['mpg'] > upper_bound)]
print(outliers[['car name', 'mpg', 'model year']])
```
