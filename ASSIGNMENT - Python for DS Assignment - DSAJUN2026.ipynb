```python
# Perform meaningful visualizations using the dataset. Bring at least 3 meaningful visualizations
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read Excel file
xls = pd.ExcelFile('iris.xls')
df = pd.read_excel('iris.xls')

# Set visual style
sns.set_theme(style="whitegrid")

# Visualization 1: Scatter Plot (PL vs PW colored by Classification)
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='PL',
    y='PW',
    hue='Classification',
    style='Classification',
    s=75,
    palette='Set2'
)
plt.title('Visualization 1: Petal Length vs. Petal Width by Species', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Petal Length (cm)', fontsize=11)
plt.ylabel('Petal Width (cm)', fontsize=11)
plt.legend(title='Classification', frameon=True)
plt.tight_layout()
plt.show()

# Visualization 2: Box Plots for Feature Distributions
plt.figure(figsize=(10, 6))
df_melted = pd.melt(
    df,
    id_vars=['Classification'],
    value_vars=['SL', 'SW', 'PL', 'PW'],
    var_name='Feature',
    value_name='Measurement'
)
sns.boxplot(
    data=df_melted,
    x='Feature',
    y='Measurement',
    hue='Classification',
    palette='Pastel1'
)
plt.title('Visualization 2: Distribution of Features Across Species', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Feature (SL: Sepal Length, SW: Sepal Width, PL: Petal Length, PW: Petal Width)', fontsize=11)
plt.ylabel('Measurement (cm)', fontsize=11)
plt.legend(title='Classification', frameon=True)
plt.tight_layout()
plt.show()

# Visualization 3: Correlation Heatmap
plt.figure(figsize=(7, 5))
corr = df[['SL', 'SW', 'PL', 'PW']].corr()
sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=1,
    square=True,
    cbar_kws={'shrink': 0.8}
)
plt.title('Visualization 3: Feature Correlation Matrix', fontsize=13, fontweight='bold', pad=12)
plt.tight_layout()
plt.show()
```
