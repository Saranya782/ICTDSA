Bank Account Manager
a. Simulate a bank account with
deposit, withdrawal, and balance
enquiry operations
b. Use variables and data types to store account information
c. Implement conditional statements to check for sufficient balance
before withdrawal
d. Use functions to perform deposit, withdrawal, and balance inquiry
operations.

```python
#Bank Account Manager
acc_balance=0.0

def deposite(amount):
    global acc_balance
    if amount>0:
      acc_balance=acc_balance+amount
      print(f"{amount} deposited successfully")
    else:
      print("Check amount")

def withdraw(amount):
  global acc_balance
  if amount<=0:
    print("Invalid")
  elif amount>acc_balance:
    print("insufficient bank balance")
  else:
    acc_balance=acc_balance-amount
    print(f"{amount} Withdrawn Successfully")

def balance():
   print(f"Current balance:{acc_balance}")

deposite(5000)
withdraw(3500)
balance()
```

Weather Data Analyzer
a. Analyze weather data (temperature, humidity, etc.) for a given region
b. Use variables and data types to store weather data
c. Implement conditional statements to determine weather conditions(sunny, rainy, etc.)
d. Use functions to calculate average temperature and humidity, and print results

```python
#Weather Data Analyzer
def avg(readings):
  total=sum(readings)
  count=len(readings)
  return total / count

def analyze_weather():
  region="Trivandrum"
  temp=[28, 24, 25, 26, 25, 24]
  humidity=[75, 84, 82, 83, 80, 84]

  avg_temp=avg(temp)
  avg_humidity=avg(humidity)
  weather_condition=""
  if avg_humidity>80:
    weather_condition="Rainy"
  elif avg_temp>25:
    weather_condition="Sunny"
  elif avg_temp<10:
    weather_condition="Cold"
  else:
    weather_condition="Cloudy"

  print(f"Region: {region}")
  print(f"Average Temperature: {avg_temp} C")
  print(f"Average Humidity: {avg_humidity} %")
  print(f"Weather Condition: {weather_condition}")

analyze_weather()
```
